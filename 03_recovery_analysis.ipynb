```python
# STEP 12C — Overall recovery funnel

targeting = pd.read_csv(
    CLEAN_DIR / "daily_targeting_clean.csv"
)

calls = pd.read_csv(
    CLEAN_DIR / "calls_clean.csv"
)

ptp = pd.read_csv(
    CLEAN_DIR / "promises_to_pay_clean.csv"
)

payments = pd.read_csv(
    CLEAN_DIR / "payments_clean.csv"
)

# Unique account populations
targeted = targeting["account_id"].dropna().nunique()
called = calls["account_id"].dropna().nunique()
ptp_accounts = ptp["account_id"].dropna().nunique()

successful_payment_accounts = payments.loc[
    payments["payment_status"] == "SUCCESS",
    "account_id"
].dropna().nunique()

funnel = pd.DataFrame({
    "stage": [
        "Targeted Accounts",
        "Called Accounts",
        "PTP Accounts",
        "Paying Accounts"
    ],
    "accounts": [
        targeted,
        called,
        ptp_accounts,
        successful_payment_accounts
    ]
})

funnel["pct_of_targeted"] = (
    funnel["accounts"] / targeted * 100
).round(2)

display(funnel)
```

```python
# STEP 12D — Correct targeted-account funnel

targeted_accounts = set(
    targeting["account_id"].dropna().unique()
)

called_accounts = set(
    calls["account_id"].dropna().unique()
)

ptp_accounts_set = set(
    ptp["account_id"].dropna().unique()
)

paying_accounts_set = set(
    payments.loc[
        payments["payment_status"] == "SUCCESS",
        "account_id"
    ].dropna().unique()
)

# Keep only accounts that belong to the targeted population
called_targeted = targeted_accounts & called_accounts
ptp_targeted = targeted_accounts & ptp_accounts_set
paying_targeted = targeted_accounts & paying_accounts_set

corrected_funnel = pd.DataFrame({
    "stage": [
        "Targeted Accounts",
        "Called Targeted Accounts",
        "PTP Targeted Accounts",
        "Paying Targeted Accounts"
    ],
    "accounts": [
        len(targeted_accounts),
        len(called_targeted),
        len(ptp_targeted),
        len(paying_targeted)
    ]
})

corrected_funnel["pct_of_targeted"] = (
    corrected_funnel["accounts"]
    / len(targeted_accounts)
    * 100
).round(2)

display(corrected_funnel)
```

```python
# STEP 12E — Contact Rate and RPC

dispositions = pd.read_csv(
    RAW_DIR / "call_dispositions.csv"
)

# Accounts that were targeted
targeted_accounts = set(
    targeting["account_id"].dropna().unique()
)

# Keep only disposition records for targeted accounts
targeted_dispositions = dispositions[
    dispositions["account_id"].isin(targeted_accounts)
].copy()

# Inspect disposition codes
print("Disposition codes:")
display(
    targeted_dispositions["disposition_code"]
    .value_counts()
    .to_frame("count")
)

# Conservative contact definition:
# A successful contact is a disposition that indicates
# an actual interaction with the borrower.
contact_codes = [
    "PROMISE_TO_PAY",
    "PTP",
    "CALLBACK",
    "PAID",
    "DISPUTE",
    "REFUSED",
    "PTP_BROKEN"
]

contacted_accounts = targeted_dispositions[
    targeted_dispositions["disposition_code"].isin(contact_codes)
]["account_id"].nunique()

targeted_count = len(targeted_accounts)

contact_rate = contacted_accounts / targeted_count * 100

print("Targeted accounts:", targeted_count)
print("Contacted accounts:", contacted_accounts)
print(f"Contact Rate: {contact_rate:.2f}%")
```

```python
# STEP 12F — Inspect likely RPC-related outcomes

rpc_candidates = [
    "PROMISE_TO_PAY",
    "PTP",
    "PTP_BROKEN",
    "PAID",
    "DISPUTE",
    "CALLBACK",
    "REFUSED"
]

rpc_review = (
    targeted_dispositions[
        targeted_dispositions["disposition_code"].isin(rpc_candidates)
    ]
    .groupby("disposition_code")["account_id"]
    .nunique()
    .reset_index(name="unique_accounts")
    .sort_values("unique_accounts", ascending=False)
)

display(rpc_review)

print("\nDisposition versions for these outcomes:")
display(
    pd.crosstab(
        targeted_dispositions[
            targeted_dispositions["disposition_code"].isin(rpc_candidates)
        ]["disposition_code"],
        targeted_dispositions[
            targeted_dispositions["disposition_code"].isin(rpc_candidates)
        ]["disposition_version"]
    )
)
```

```python
# STEP 12G — PTP Rate and PTP Kept Rate

# Targeted account population
targeted_accounts = set(
    targeting["account_id"].dropna().unique()
)

# Accounts with a PTP
ptp_targeted = ptp[
    ptp["account_id"].isin(targeted_accounts)
].copy()

ptp_accounts = ptp_targeted["account_id"].nunique()

# Contacted accounts from our working contact definition
contacted_accounts = targeted_dispositions[
    targeted_dispositions["disposition_code"].isin(contact_codes)
]["account_id"].nunique()

# PTP Rate
ptp_rate_on_targeted = (
    ptp_accounts / len(targeted_accounts) * 100
)

ptp_rate_on_contacted = (
    ptp_accounts / contacted_accounts * 100
)

# Actual PTP status counts
kept_ptp_accounts = ptp_targeted.loc[
    ptp_targeted["status"] == "KEPT",
    "account_id"
].nunique()

ptp_kept_rate = (
    kept_ptp_accounts / ptp_accounts * 100
)

print("Targeted accounts:", len(targeted_accounts))
print("Contacted accounts:", contacted_accounts)
print("PTP accounts:", ptp_accounts)
print("KEPT PTP accounts:", kept_ptp_accounts)

print(
    f"\nPTP Rate among targeted accounts: "
    f"{ptp_rate_on_targeted:.2f}%"
)

print(
    f"PTP Rate among contacted accounts: "
    f"{ptp_rate_on_contacted:.2f}%"
)

print(
    f"PTP Kept Rate: "
    f"{ptp_kept_rate:.2f}%"
)
```

```python
# STEP 13 — February vs March DPD mix

accounts = pd.read_csv(
    CLEAN_DIR / "accounts_clean.csv"
)

accounts["dpd_bucket"] = pd.cut(
    accounts["dpd"],
    bins=[-1, 0, 30, 60, 90, float("inf")],
    labels=[
        "0",
        "1-30",
        "31-60",
        "61-90",
        "90+"
    ]
)

# Paying accounts in February and March
payments["month"] = pd.to_datetime(
    payments["event_at"],
    errors="coerce"
).dt.to_period("M").astype(str)

success = payments[
    payments["payment_status"] == "SUCCESS"
].copy()

success = success.merge(
    accounts[["account_id", "dpd_bucket"]],
    on="account_id",
    how="left"
)

dpd_mix = (
    success[
        success["month"].isin(["2026-02", "2026-03"])
    ]
    .groupby(["month", "dpd_bucket"])
    .agg(
        recovery_amount=("amount", "sum"),
        paying_accounts=("account_id", "nunique")
    )
    .reset_index()
)

dpd_mix["recovery_share_pct"] = (
    dpd_mix["recovery_amount"]
    / dpd_mix.groupby("month")["recovery_amount"]
      .transform("sum")
    * 100
)

display(dpd_mix)
```

```python
# STEP 13B — Recovery per paying account by DPD

dpd_efficiency = (
    success[
        success["month"].isin(["2026-02", "2026-03"])
    ]
    .groupby(["month", "dpd_bucket"])
    .agg(
        recovery_amount=("amount", "sum"),
        paying_accounts=("account_id", "nunique")
    )
    .reset_index()
)

dpd_efficiency["recovery_per_paying_account"] = (
    dpd_efficiency["recovery_amount"]
    / dpd_efficiency["paying_accounts"]
)

display(dpd_efficiency)
```

```python
# STEP 13C — February vs March channel mix

targeting["target_date"] = pd.to_datetime(
    targeting["target_date"],
    errors="coerce"
)

targeting["month"] = (
    targeting["target_date"]
    .dt.to_period("M")
    .astype(str)
)

channel_mix = (
    targeting[
        targeting["month"].isin(["2026-02", "2026-03"])
    ]
    .groupby(["month", "recommended_channel"])
    .agg(
        targeted_accounts=("account_id", "nunique")
    )
    .reset_index()
)

channel_mix["share_pct"] = (
    channel_mix["targeted_accounts"]
    / channel_mix.groupby("month")["targeted_accounts"]
      .transform("sum")
    * 100
)

display(channel_mix)
```

```python
# STEP 13D — Recovery by targeting channel
# Fixed version: uses target_month to avoid month-column conflicts

# Reload clean targeting
targeting = pd.read_csv(
    CLEAN_DIR / "daily_targeting_clean.csv"
)

# Create target month
targeting["target_date"] = pd.to_datetime(
    targeting["target_date"],
    errors="coerce"
)

targeting["target_month"] = (
    targeting["target_date"]
    .dt.to_period("M")
    .astype(str)
)

# Keep February and March targeting records
targeting_feb_mar = (
    targeting[
        targeting["target_month"].isin(["2026-02", "2026-03"])
    ][
        ["account_id", "target_month", "recommended_channel"]
    ]
    .drop_duplicates()
)

# Make sure successful payments have a clean payment month
success_for_channel = success.copy()

success_for_channel["payment_month"] = (
    pd.to_datetime(
        success_for_channel["event_at"],
        errors="coerce"
    )
    .dt.to_period("M")
    .astype(str)
)

# Merge payments with targeting
success_channel = success_for_channel.merge(
    targeting_feb_mar,
    on="account_id",
    how="inner"
)

# Keep only payments in the same month as the targeting record
success_channel = success_channel[
    success_channel["payment_month"]
    == success_channel["target_month"]
].copy()

# Group by target month and channel
channel_recovery = (
    success_channel
    .groupby(
        ["target_month", "recommended_channel"],
        as_index=False
    )
    .agg(
        recovery_amount=("amount", "sum"),
        paying_accounts=("account_id", "nunique")
    )
)

channel_recovery["recovery_per_paying_account"] = (
    channel_recovery["recovery_amount"]
    / channel_recovery["paying_accounts"]
)

channel_recovery["recovery_share_pct"] = (
    channel_recovery["recovery_amount"]
    /
    channel_recovery.groupby("target_month")["recovery_amount"]
    .transform("sum")
    * 100
)

display(channel_recovery)
```

```python
# STEP 13F — February vs March risk-segment mix

risk_recovery = (
    success_portfolio[
        success_portfolio["month"].isin(["2026-02", "2026-03"])
    ]
    .groupby(
        ["month", "risk_segment"],
        as_index=False
    )
    .agg(
        recovery_amount=("amount", "sum"),
        paying_accounts=("account_id", "nunique")
    )
)

risk_recovery["recovery_share_pct"] = (
    risk_recovery["recovery_amount"]
    /
    risk_recovery.groupby("month")["recovery_amount"]
    .transform("sum")
    * 100
)

risk_recovery["recovery_per_paying_account"] = (
    risk_recovery["recovery_amount"]
    / risk_recovery["paying_accounts"]
)

display(risk_recovery)
```

```python
# STEP 15 — February vs March targeted population

targeting_counts = (
    targeting[
        targeting["target_month"].isin(["2026-02", "2026-03"])
    ]
    .groupby("target_month")["account_id"]
    .nunique()
    .reset_index(name="targeted_accounts")
)

targeting_counts["change_pct"] = (
    targeting_counts["targeted_accounts"]
    .pct_change() * 100
)

display(targeting_counts)
```

```python
# STEP 16 — Target-to-payment conversion

feb_targeted = targeting_counts.loc[
    targeting_counts["target_month"] == "2026-02",
    "targeted_accounts"
].iloc[0]

mar_targeted = targeting_counts.loc[
    targeting_counts["target_month"] == "2026-03",
    "targeted_accounts"
].iloc[0]

feb_paying = feb["paying_accounts"]
mar_paying = mar["paying_accounts"]

feb_conversion = feb_paying / feb_targeted * 100
mar_conversion = mar_paying / mar_targeted * 100

conversion_change = mar_conversion - feb_conversion

print(f"February target-to-payment conversion: {feb_conversion:.2f}%")
print(f"March target-to-payment conversion:    {mar_conversion:.2f}%")
print(f"Change:                                {conversion_change:+.2f} percentage points")
```

```python
# STEP 17 — Monthly target-to-payment conversion

# Monthly targeted accounts
monthly_targeted = (
    targeting
    .groupby("target_month")["account_id"]
    .nunique()
    .reset_index(name="targeted_accounts")
)

# Monthly successful paying accounts
success_monthly = (
    success
    .groupby("month")["account_id"]
    .nunique()
    .reset_index(name="paying_accounts")
)

# Merge
monthly_conversion = monthly_targeted.merge(
    success_monthly,
    left_on="target_month",
    right_on="month",
    how="left"
)

monthly_conversion["paying_accounts"] = (
    monthly_conversion["paying_accounts"]
    .fillna(0)
)

monthly_conversion["target_to_payment_conversion_pct"] = (
    monthly_conversion["paying_accounts"]
    / monthly_conversion["targeted_accounts"]
    * 100
)

display(
    monthly_conversion[
        [
            "target_month",
            "targeted_accounts",
            "paying_accounts",
            "target_to_payment_conversion_pct"
        ]
    ]
)
```

```python
# STEP 18A — Overall channel performance

channel_overall = (
    success_channel
    .groupby("recommended_channel", as_index=False)
    .agg(
        recovery_amount=("amount", "sum"),
        paying_accounts=("account_id", "nunique")
    )
)

channel_overall["recovery_per_paying_account"] = (
    channel_overall["recovery_amount"]
    / channel_overall["paying_accounts"]
)

channel_overall["recovery_share_pct"] = (
    channel_overall["recovery_amount"]
    / channel_overall["recovery_amount"].sum()
    * 100
)

display(
    channel_overall.sort_values(
        "recovery_amount",
        ascending=False
    )
)
```

```python
# STEP 18B — Channel target-to-payment conversion

channel_targeted = (
    targeting[
        targeting["target_month"].isin([
            "2026-01",
            "2026-02",
            "2026-03",
            "2026-04",
            "2026-05",
            "2026-06",
            "2026-07"
        ])
    ]
    .groupby("recommended_channel")["account_id"]
    .nunique()
    .reset_index(name="targeted_accounts")
)

channel_paid = (
    success_channel
    .groupby("recommended_channel")["account_id"]
    .nunique()
    .reset_index(name="paying_accounts")
)

channel_conversion = channel_targeted.merge(
    channel_paid,
    on="recommended_channel",
    how="left"
)

channel_conversion["paying_accounts"] = (
    channel_conversion["paying_accounts"]
    .fillna(0)
)

channel_conversion["target_to_payment_pct"] = (
    channel_conversion["paying_accounts"]
    / channel_conversion["targeted_accounts"]
    * 100
)

display(
    channel_conversion.sort_values(
        "target_to_payment_pct",
        ascending=False
    )
)
```

```python
# STEP 18D — Agent hours by channel

agent_sessions = pd.read_csv(
    RAW_DIR / "agent_sessions.csv"
)

agent_sessions["login_at"] = pd.to_datetime(
    agent_sessions["login_at"],
    errors="coerce"
)

agent_sessions["logout_at"] = pd.to_datetime(
    agent_sessions["logout_at"],
    errors="coerce"
)

# Session duration in hours
agent_sessions["session_hours"] = (
    agent_sessions["logout_at"]
    - agent_sessions["login_at"]
).dt.total_seconds() / 3600

print("Total sessions:", len(agent_sessions))

print(
    "Missing session durations:",
    agent_sessions["session_hours"].isna().sum()
)

print(
    "Negative session durations:",
    (agent_sessions["session_hours"] < 0).sum()
)

channel_hours = (
    agent_sessions
    .groupby("channel", as_index=False)
    .agg(
        total_agent_hours=("session_hours", "sum"),
        sessions=("session_id", "nunique")
    )
)

display(channel_hours)
```

```python
# STEP 18E — Recovery per agent-hour by channel

recovery_by_channel = (
    success_channel
    .groupby("recommended_channel", as_index=False)
    .agg(
        recovery_amount=("amount", "sum")
    )
)

recovery_by_channel = recovery_by_channel.merge(
    channel_hours,
    left_on="recommended_channel",
    right_on="channel",
    how="left"
)

recovery_by_channel["recovery_per_agent_hour"] = (
    recovery_by_channel["recovery_amount"]
    / recovery_by_channel["total_agent_hours"]
)

display(
    recovery_by_channel[
        [
            "recommended_channel",
            "recovery_amount",
            "total_agent_hours",
            "recovery_per_agent_hour"
        ]
    ].sort_values(
        "recovery_per_agent_hour",
        ascending=False
    )
)
```

```python
# STEP 18F — Channel efficiency opportunity scenario

best_efficiency = recovery_by_channel["recovery_per_agent_hour"].max()
worst_efficiency = recovery_by_channel["recovery_per_agent_hour"].min()

efficiency_gap = best_efficiency - worst_efficiency
relative_advantage = (
    best_efficiency / worst_efficiency - 1
) * 100

print(f"Best recovery/agent-hour:  ₹{best_efficiency:,.2f}")
print(f"Worst recovery/agent-hour: ₹{worst_efficiency:,.2f}")
print(f"Efficiency gap:             ₹{efficiency_gap:,.2f}")
print(f"Relative advantage:         {relative_advantage:.2f}%")
```
